```python
import pandas as pd
df=pd.read_csv('kidney_disease.csv')
```

```python
df.info()
```

```python
df.head()
```

```python
df['age'].fillna(df['age'].mean(),inplace=True)
```

```python
df['pc'].fillna(df['pc'].mode()[0],inplace=True)
```

```python
df['bp'].fillna(df['bp'].median(),inplace=True)
```

```python
df['htn'].fillna('no',inplace=True)
```

```python
df.drop('rc',axis=1,inplace=True)
```

# Data Preprocessing + for loop

```python
df.isnull().sum()
```

```python
for col in df.columns:
    if df[col].isnull().any():
        if df[col].dtype == 'object':
            df[col].fillna(df[col].mode()[0], inplace=True)
        else:
            df[col].fillna(df[col].median(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df.info()
```

# Encoding + For loop

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder=LabelEncoder()
```

```python
df['rbc']=encoder.fit_transform(df['rbc'])
```

```python
df.nunique()
```

```python
dummies=pd.get_dummies(df['rbc'],prefix='col',dtype=int)
```

```python
dummies
```

```python
df=pd.concat([df.drop(columns=['rbc']),dummies],axis=1)
```

```python
for col in df.columns:
    if df[col].dtype=='object':
        if df[col].nunique()<=5:
            dummies=pd.get_dummies(df[col],prefix=col,dtype=int)
            df=pd.concat([df.drop(columns=col),dummies],axis=1)
       
        else:
            df[col]=encoder.fit_transform(df[col])
        
```
