Agar berilgan so’zning bosh xarfi ‘a’ bolsa yaxshi aks xolda yomon degan sozni ekrang chiqaring 

```python
a = 'hello'
if a[0] == 'a':
    print('yaxshi')
else:
    print('yomon')
```

Agar berilgan so’zning oxirgi xarfi ‘a’ bolsa yaxshi aks xolda yomon degan sozni ekrang chiqaring 

```python
if a[-1] == 'a':
    print('yaxshi')
else:
    print('yomon')
```

Agar berilgan so’zning bosh xarfi ‘b’ va oxirgi xarfi ‘a’ bolsa yaxshi aks xolda yomon degan sozni ekrang chiqaring 

```python
if a[0] == 'b' and a[-1] == 'a':
    print('yaxshi')
else:
    print('yomon')
```

```python
import pandas as pd
df=pd.read_csv('Loan-Approval-Prediction.csv')
```

Mean orqali to’ldiring

```python
df['LoanAmount'].fillna(df['LoanAmount'].mean(),inplace=True)
```

Mode orqali to’ldiring

```python
df['Self_Employed'].fillna(df['Self_Employed'].mode()[0],inplace=True)
```

Median orqali to’ldiring

```python
df['Loan_Amount_Term'].fillna(df['Loan_Amount_Term'].median(),inplace=True)
```

Fixed qiymat bn to’ldiring

```python
df['Credit_History'].fillna('1.0',inplace=True)
```
